```python
import geopandas as gpd
import pandas as pd

# Cargar el GeoJSON original desde data/raw
path_raw = '../data/raw/desercion_upl.geojson'
gdf = gpd.read_file(path_raw)

# Verificar dimensiones y columnas disponibles
print("Dimensiones del dataset:", gdf.shape)
print("Columnas encontradas:", gdf.columns.tolist())
gdf.head()
```

```python
# Convertir coordenadas al sistema estándar WGS84 (EPSG:4326) para visores web
gdf = gdf.to_crs(epsg=4326)

# Rellenar valores nulos en columnas numéricas si existen
cols_numericas = gdf.select_dtypes(include=['float64', 'int64']).columns
gdf[cols_numericas] = gdf[cols_numericas].fillna(0)

# Guardar el GeoJSON procesado en la carpeta data/processed
path_processed = '../data/processed/desercion_upl_procesado.geojson'
gdf.to_file(path_processed, driver='GeoJSON')
print("✅ GeoJSON procesado guardado con éxito en:", path_processed)
```
